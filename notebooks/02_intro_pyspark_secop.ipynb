```python
import pandas as pd
import requests

url = "https://www.datos.gov.co/api/v3/views/jbjy-vk9h/query.json"
consulta = {
    "query": "SELECT *",
    "page": {
        "pageNumber": 1,
        "pageSize": 10
    },
    "includeSynthetic": False
}

respuesta = requests.post(url, json=consulta)
respuesta.raise_for_status()
tabla = pd.DataFrame(respuesta.json())
tabla.head()
```

```python
from pyspark.sql import SparkSession

# Iniciar una sesión de spark
spark= SparkSession.builder \
    .appName("Secop Data Analysis") \
    .getOrCreate()

# Creamos una tabla spark a partir del dataframe
spark_df = spark.createDataFrame(tabla)

# Veamos una muestra de la tabla
spark_df.show(5)
```

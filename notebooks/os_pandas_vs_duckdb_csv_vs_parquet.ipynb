```python
import pandas as pd
import time

print("Iniciando lectura pandas")

try:
    start_time = time.time()
    df = pd.read_csv(f"{path}/train.csv")
    media = df["fare_amount"].mean()
    end_time = time.time()
    print(f"Tiempo de lectura pandas: {end_time - start_time:.2f} segundos")
except Exception as e:
    print(f"Errores al leer el archivo: {e}")
```

```python
## Leyendo con DuckDB

# DuckDB optimiza la RAM con el paradigma Out of Core (Toma los datos y va cargando lo que se vaya neecesitando en la RAM, aprovehcando los recursos disponibles de un solo nodo) 
```

```python
import duckdb 
import time
path = "/home/vscode/.cache/kagglehub/competitions/new-york-city-taxi-fare-prediction"

print("Iniciando lectura duckdb")

try:
    start_time = time.time()
    df = pd.read_csv(f"{path}/train.csv")
    media = df["fare_amount"].mean()
    end_time = time.time()
    print(f"Tiempo de lectura pandas: {end_time - start_time:.2f} segundos")
except Exception as e:
    print(f"Errores al leer el archivo: {e}")
```
